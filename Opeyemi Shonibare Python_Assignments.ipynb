## Assignment 2: Data Types Practice

```python
student_name = "Opeyemi"
age = 27
cgpa = 3.20
is_graduated = True
ai_tools = ["ChatGPT", "Gemini", "Claude", "Copilot", "Perplexity"]
programming_languages = ("Python", "Java", "C++")
student_info = {"name": "Opeyemi", "age": 27, "state": "Lagos"}

print(student_name, type(student_name))
print(age, type(age))
print(cgpa, type(cgpa))
print(is_graduated, type(is_graduated))
print(ai_tools, type(ai_tools))
print(programming_languages, type(programming_languages))
print(student_info, type(student_info))
```

## Assignment 3: String Operations

```python
text = "   Artificial Intelligence   "

print("Original String:", text)
stripped_text = text.strip()
print("After Strip:", stripped_text)
print("Uppercase:", stripped_text.upper())
print("Lowercase:", stripped_text.lower())
print("Sliced String:", stripped_text[0:10])
new_text = stripped_text + " is transforming the world."
print("Concatenated String:", new_text)
```
