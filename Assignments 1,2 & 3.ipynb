```python
# Assignment 2: Data Types Practice

# String
string_var = "Artificial Intelligence"

# Integer
int_var = 25

# Float
float_var = 3.75

# Boolean
bool_var = True

# List (5 AI tools)
list_var = ["ChatGPT", "TensorFlow", "PyTorch", "Scikit-learn", "Keras"]

# Tuple
tuple_var = ("AI", "Machine Learning", "Deep Learning")

# Dictionary
dict_var = {
    "name": "Philip",
    "state": "Ogun",
    "field": "Engineering"
}

print("=== Data Types ===")

print(string_var, type(string_var))
print(int_var, type(int_var))
print(float_var, type(float_var))
print(bool_var, type(bool_var))
print(list_var, type(list_var))
print(tuple_var, type(tuple_var))
print(dict_var, type(dict_var))
```

```python
# Assignment 3: String Operations

text = "  python for ai builders  "

print("\n=== String Operations ===")

# Uppercase
print("Uppercase:", text.upper())

# Lowercase
print("Lowercase:", text.lower())

# Strip (remove spaces)
print("Stripped:", text.strip())

# Slicing
print("Sliced:", text[2:10])

# Concatenation
new_text = text.strip() + " - Bootcamp"
print("Concatenated:", new_text)
```
