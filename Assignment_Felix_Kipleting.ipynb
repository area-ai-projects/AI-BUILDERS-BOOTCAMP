```python
#ASSIGNMENT 2: DATA TYPES
#Objectives: Understanding Python Data types
```

```python
# String
my_string = "Artificial Intelligence"

# Integer
my_integer = 25

# Float
my_float = 3.14

# Boolean
my_boolean = True

# List (5 AI tools)
ai_tools = ["ChatGPT", "Gemini", "Claude", "Copilot", "Perplexity"]

# Tuple
my_tuple = ("Python", "Machine Learning", "Deep Learning")

# Dictionary
my_dict = {
    "name": "Felix",
    "age": 24,
    "course": "Geospatial Engineering"
}

print("String:", my_string, type(my_string))
print("Integer:", my_integer, type(my_integer))
print("Float:", my_float, type(my_float))
print("Boolean:", my_boolean, type(my_boolean))
print("List:", ai_tools, type(ai_tools))
print("Tuple:", my_tuple, type(my_tuple))
print("Dictionary:", my_dict, type(my_dict))
```

```python
#ASSIGNMENT 3: STRING OPERATIONS
#Objective: To understand string manipulation 
```

```python
# Given string
text = "  Artificial Intelligence is the future of technology  "

# Uppercase (capital letters)
print("Uppercase:", text.upper())

# Lowercase (small letters)
print("Lowercase:", text.lower())

# Slicing (first 10 characters)
print("Slicing:", text[0:10])

# Strip (remove spaces)
print("Strip:", text.strip())

# Concatenation ( combine texts)
new_text = text.strip() + " and data science is powerful"
print("Concatenation:", new_text)
```
