# Assignment 2: Data Types Practice

```python
print("ASSIGNMENT 2: DATA TYPES PRACTICE")
print("-" * 40)

# String
text = "Artificial Intelligence"

# Integer
number = 100

# Float
price = 99.99

# Boolean
is_learning_python = True

# List
ai_tools = [
    "ChatGPT",
    "Claude",
    "Gemini",
    "Copilot",
    "Perplexity"
]

# Tuple
programming_languages = (
    "Python",
    "Java",
    "JavaScript"
)

# Dictionary
student = {
    "name": "Abdulrahman",
    "age": 20,
    "state": "Bauchi"
}

# Print values and types
print("Value:", text)
print("Type:", type(text))

print("\nValue:", number)
print("Type:", type(number))

print("\nValue:", price)
print("Type:", type(price))

print("\nValue:", is_learning_python)
print("Type:", type(is_learning_python))

print("\nValue:", ai_tools)
print("Type:", type(ai_tools))

print("\nValue:", programming_languages)
print("Type:", type(programming_languages))

print("\nValue:", student)
print("Type:", type(student))

print("\n")


```

# Assignment 3: String Operations

```python
print("ASSIGNMENT 3: STRING OPERATIONS")
print("-" * 40)

text = "  Welcome to AREA AI Builders Bootcamp  "

print("Original String:")
print(text)

# Uppercase
print("\nUppercase:")
print(text.upper())

# Lowercase
print("\nLowercase:")
print(text.lower())

# Slicing
print("\nSlicing (first 7 characters):")
print(text[:7])

# Strip
print("\nStrip:")
print(text.strip())

# Concatenation
first_part = "Python "
second_part = "Fundamentals"

print("\nConcatenation:")
print(first_part + second_part)

print("\nEnd of Assignment")
```
