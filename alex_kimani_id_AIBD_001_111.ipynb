```python
# Assignment 3: String Operations

text = input("Enter a message for the AREA AI system: ")

# Strip (remove extra spaces)
clean_text = text.strip()

# Uppercase
print("\nUppercase:", clean_text.upper())

# Lowercase
print("Lowercase:", clean_text.lower())

# Slicing
print("First 10 characters:", clean_text[:10])
print("Last 5 characters:", clean_text[-5:])

# Concatenation
final_message = "AREA AI System received: " + clean_text
print(final_message)
```
